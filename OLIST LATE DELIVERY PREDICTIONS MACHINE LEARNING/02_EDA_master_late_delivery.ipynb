```python
num_cols= m_df.select_dtypes(include=['int64','float64']).columns.tolist()
print(num_cols)
                             
```

```python
outlier_summary = []
for col in num_cols:
    Q1 = m_df[col].quantile(0.25)
    Q3 = m_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers =((m_df[col] < lower) | (m_df[col] > upper)).sum()
    outlier_percent = (outliers / len(m_df)) * 100
    outlier_summary.append({                                                        
    'Feature': col,                                                                 
    'outlier_count':outliers,                                                       
    'Percentage of Outliers': round(outlier_percent, 2)                             
    })                                                                              
    outlier_df = pd.DataFrame(outlier_summary)                                         
    outlier_df = outlier_df.sort_values(by = 'Percentage of Outliers', ascending=False)
```
